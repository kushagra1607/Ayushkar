# Title

Mean-Field Lifshitz Line and Spinodal of Polydisperse, Conformationally Asymmetric A/B/A–B Ternary Polymer Blends from the Random Phase Approximation

# Subproblem 1

## Prompt

Consider an incompressible melt containing three species of flexible, continuous Gaussian chains built from two monomer types, A and B. All segments have the same volume. The two monomer types have different statistical segment lengths, $b_A$ and $b_B$ (conformational asymmetry).

The melt contains:
- **A–B diblock copolymer**, volume fraction $\phi_C=1-\phi_A-\phi_B$.
- **A homopolymer**, volume fraction $\phi_A$.
- **B homopolymer**, volume fraction $\phi_B$.

All chain lengths are **polydisperse**. Every A sequence (the A block of the diblock and the A homopolymer) has a Schulz–Zimm length distribution with dispersity $Đ_A=M_w/M_n$. Every B sequence has a Schulz–Zimm distribution with dispersity $Đ_B$. Both dispersities lie in $[1,3]$. The A and B block lengths within a diblock are statistically independent.

The number-average lengths are:
- A block: $f\bar N$; B block: $(1-f)\bar N$. Here $\bar N$ is the number-average diblock length.
- A homopolymer: $\alpha_A\bar N$; B homopolymer: $\alpha_B\bar N$.

A Schulz–Zimm distribution with number-average $\bar n$ and dispersity $Đ$ has number distribution $P(n)\propto n^{k-1}e^{-kn/\bar n}$ with $k=1/(Đ-1)$. The case $Đ=1$ is the monodisperse limit.

Choose a reference length $b_0$ and define $x=q^2\bar Nb_0^2/6$, $\beta_A=(b_A/b_0)^2$, and $\beta_B=(b_B/b_0)^2$.

Let $S^0_{ij}(q)$ ($i,j\in\{A,B\}$) be the ideal (non-interacting) segment–segment correlation functions per segment of the melt:
$$S^0_{ij}(q)=\frac{1}{M}\sum_{\text{chains}}\;\sum_{m\in i,\;n\in j}\left\langle e^{i\mathbf q\cdot(\mathbf r_m-\mathbf r_n)}\right\rangle_0,$$
where $M$ is the total number of segments, both segment indices run over the same chain, and the sum over chains includes the length distributions. The reduced functions $s_{ij}(x)=S^0_{ij}(q)/\bar N$ and their $x$-derivatives must be evaluated to relative accuracy $10^{-12}$ and $10^{-11}$, respectively, for every $x\ge0$ and every $Đ_A,Đ_B\in[1,3]$, including $Đ$ exactly $1$ and arbitrarily close to $1$.

A and B segments interact through a Flory–Huggins parameter $\chi$. The scattering function $S(q)$ of the homogeneous disordered melt is given by the incompressible two-component random phase approximation (de Gennes/Leibler) built from these $S^0_{ij}$.

For a given $\chi\bar N$ below the spinodal (so $S(q)$ is finite and positive for all $q$), characterize the scattering peak of $I(x)=S(q)/\bar N$ on $x\in[0,50]$. Return:
- `x_pk`: the location of the **global** maximum of $I$ on $[0,50]$. Return exactly `0.0` if the maximum is at $q=0$.
- `I_pk`: the peak height $I(x_{pk})$.
- `x_lo`: the half-maximum point nearest the peak on its low-$q$ side, i.e. the largest $x<x_{pk}$ with $I(x)=I_{pk}/2$. Return `nan` if $I(x)>I_{pk}/2$ for all $x\in[0,x_{pk})$, including when $x_{pk}=0$.
- `x_hi`: the half-maximum point nearest the peak on its high-$q$ side, i.e. the smallest $x>x_{pk}$ with $I(x)=I_{pk}/2$. It exists in $(x_{pk},50]$ for all inputs considered.

Requirements:
- Return four Python floats.
- Accuracy: relative $10^{-10}$ in `x_pk`, and relative $10^{-11}$ in `I_pk`, `x_lo` and `x_hi`.
- Inputs always contain both A and B segments.
- Only the function `scattering_peak` itself is kept when your solution is executed. Any helper functions and imports must be defined **inside its body**.

## Background

**Ideal correlation functions.** For a continuous Gaussian sequence, averaging $e^{-un}$ over a Schulz–Zimm length distribution gives $(1+\varepsilon u\bar n)^{-1/\varepsilon}$ with $\varepsilon=Đ-1$. The number average sets the chain number density $\phi_kM/\bar n_k$. With $y=u\bar n$, define
$$G_\varepsilon(y)=\frac{2[(1+\varepsilon y)^{-1/\varepsilon}-1+y]}{y^2},\qquad H_\varepsilon(y)=\frac{1-(1+\varepsilon y)^{-1/\varepsilon}}{y},$$
with $G_\varepsilon(0)=Đ$ and $H_\varepsilon(0)=1$. With $x_A=\beta_Ax$, $x_B=\beta_Bx$ and $g=1-f$:
$$s_{AA}=\phi_Cf^2G_{\varepsilon_A}(fx_A)+\phi_A\alpha_AG_{\varepsilon_A}(\alpha_Ax_A),\qquad s_{BB}=\phi_Cg^2G_{\varepsilon_B}(gx_B)+\phi_B\alpha_BG_{\varepsilon_B}(\alpha_Bx_B),$$
$$s_{AB}=\phi_Cfg\,H_{\varepsilon_A}(fx_A)\,H_{\varepsilon_B}(gx_B).$$
The cross term factorizes because the A–B path runs through the junction. Leibler's equal-$b$ shortcut is invalid here.

For stability:
- Write $(1+\varepsilon y)^{-1/\varepsilon}-1=\mathrm{expm1}(-\mathrm{log1p}(\varepsilon y)/\varepsilon)$. This is regular at $\varepsilon=0$, unlike the $k=1/(Đ-1)$ form.
- For $|y|<0.1$, use the series $T=\sum c_ny^n$, with $c_0=1$ and $c_{n+1}=-c_n(1+n\varepsilon)/(n+1)$, for $G$, $H$ and their derivatives.

**RPA scattering function.** For an incompressible two-component melt,
$$\frac{\bar N}{S(q)}=F(x)-2\chi\bar N,\qquad F=\frac{s_{AA}+s_{BB}+2s_{AB}}{s_{AA}s_{BB}-s_{AB}^2},$$
so $I(x)=1/[F(x)-2\chi\bar N]$. Below the spinodal, $F(x)>2\chi\bar N$ everywhere. For a monodisperse pure diblock, $F(0)=+\infty$ (the correlation hole) and $I(0)=0$.

**Peak.** Since $\chi$ only shifts $F$ by a constant, the peak of $I$ is at the global minimum of $F$. Its location $x_{pk}$ does not depend on $\chi\bar N$, while its height $I_{pk}=1/(F_{\min}-2\chi\bar N)$ does.

With long homopolymers or polydisperse blocks, $F$ can have a local minimum at $x=0$ and another at finite $x$, and either may be global. A global scan including $x=0$ is required.

Function-value comparisons locate a quadratic minimum only to about $\sqrt\epsilon\approx10^{-8}$ relative. To reach $10^{-10}$, solve
$$dF/dx\propto\mathcal N'\mathcal D-\mathcal N\mathcal D'=0,$$
with $\mathcal N=s_{AA}+s_{BB}+2s_{AB}$, $\mathcal D=s_{AA}s_{BB}-s_{AB}^2$, and the analytic derivatives of $s_{ij}$.

**Half-maximum points.** $I(x)=I_{pk}/2$ is equivalent to
$$F(x)=2F_{\min}-2\chi\bar N.$$
These crossings are simple roots, so a bracketing root finder gives them to machine precision.

The nearest crossing on the high-$q$ side is found by scanning upward from $x_{pk}$ to the first sign change. On the low-$q$ side, scan downward from $x_{pk}$.

On the low-$q$ side the crossing may not exist. This happens when $F(0)<2F_{\min}-2\chi\bar N$, i.e. when the scattering at $q=0$ stays above half the peak height. It is common in blends and in polydisperse melts far from the spinodal, and the half width then has to be reported from the high-$q$ side alone. As $\chi\bar N$ approaches the spinodal, the target level approaches $F_{\min}$ and both crossings close in on the peak.

## Solution

````python
import numpy as np
from scipy.optimize import brentq


def scattering_peak(f, phi_A, phi_B, alpha_A, alpha_B, beta_A, beta_B, D_A, D_B, chiN):
    """
    Disordered-state RPA scattering peak of a polydisperse, conformationally asymmetric
    A/B/A-B ternary blend: peak position, peak height and half-maximum points.

    Inputs:
      f       : float in (0, 1), number-average A fraction of the diblock
      phi_A   : float >= 0, volume fraction of A homopolymer
      phi_B   : float >= 0, volume fraction of B homopolymer (phi_A + phi_B <= 1)
      alpha_A : float > 0, number-average N_A / N
      alpha_B : float > 0, number-average N_B / N
      beta_A  : float > 0, (b_A / b0)^2
      beta_B  : float > 0, (b_B / b0)^2
      D_A     : float in [1, 3], dispersity Mw/Mn of every A sequence (A block and A homopolymer)
      D_B     : float in [1, 3], dispersity Mw/Mn of every B sequence (B block and B homopolymer)
      chiN    : float >= 0, chi * N (N = number-average diblock length), below the spinodal

    Outputs:
      x_pk : float, x = q^2 N b0^2 / 6 of the global maximum of S(q) on [0, 50] (exactly 0.0 if at q = 0)
      I_pk : float, peak height S(q_pk) / N
      x_lo : float, largest x < x_pk with S = I_pk*N/2, or nan if there is none in [0, x_pk)
      x_hi : float, smallest x > x_pk with S = I_pk*N/2
    """
    import numpy as np
    from scipy.optimize import brentq

    def _sz(y, eps):
        # Schulz-Zimm functions G, G', H, H' with eps = D - 1 >= 0:
        #   G(y) = 2[(1+eps*y)^(-1/eps) - 1 + y]/y^2,  H(y) = [1 - (1+eps*y)^(-1/eps)]/y
        y = np.asarray(y, dtype=float)
        G = np.empty_like(y); dG = np.empty_like(y)
        H = np.empty_like(y); dH = np.empty_like(y)
        sm = np.abs(y) < 0.1
        ys = y[sm]
        # series (1+eps*y)^(-1/eps) = sum_n c_n y^n, c_0 = 1, c_{n+1} = -c_n (1 + n*eps)/(n + 1)
        Gs = np.zeros_like(ys); dGs = np.zeros_like(ys)
        Hs = np.zeros_like(ys); dHs = np.zeros_like(ys)
        pw = [np.ones_like(ys)]
        for n in range(1, 82):
            pw.append(pw[-1] * ys)
        c = 1.0
        for n in range(80):
            if n >= 1:
                Hs += -c * pw[n - 1]
            if n >= 2:
                Gs += 2.0 * c * pw[n - 2]
                dHs += -(n - 1) * c * pw[n - 2]
            if n >= 3:
                dGs += 2.0 * (n - 2) * c * pw[n - 3]
            c = -c * (1.0 + n * eps) / (n + 1)
        G[sm], dG[sm], H[sm], dH[sm] = Gs, dGs, Hs, dHs
        yl = y[~sm]
        if eps == 0.0:
            T1 = np.expm1(-yl)          # (1+eps*y)^(-1/eps) - 1
            dT1 = -np.expm1(-yl)        # d/dy[(1+eps*y)^(-1/eps)] + 1
        else:
            T1 = np.expm1(-np.log1p(eps * yl) / eps)
            dT1 = -np.expm1(-(1.0 + eps) * np.log1p(eps * yl) / eps)
        Gl = 2.0 * (T1 + yl) / yl**2
        Hl = -T1 / yl
        G[~sm] = Gl
        dG[~sm] = 2.0 * dT1 / yl**2 - 2.0 * Gl / yl
        H[~sm] = Hl
        dH[~sm] = ((1.0 - dT1) - Hl) / yl
        return G, dG, H, dH

    def _corr(x_arr):
        # s_AA, s_BB, s_AB and their x-derivatives
        g = 1.0 - f
        phi_C = 1.0 - phi_A - phi_B
        eA, eB = D_A - 1.0, D_B - 1.0
        xA, xB = beta_A * x_arr, beta_B * x_arr
        G1, dG1, H1, dH1 = _sz(f * xA, eA)
        G2, dG2, _, _ = _sz(alpha_A * xA, eA)
        G3, dG3, H3, dH3 = _sz(g * xB, eB)
        G4, dG4, _, _ = _sz(alpha_B * xB, eB)
        s_AA = phi_C * f**2 * G1 + phi_A * alpha_A * G2
        s_BB = phi_C * g**2 * G3 + phi_B * alpha_B * G4
        s_AB = phi_C * f * g * H1 * H3
        ds_AA = phi_C * f**3 * beta_A * dG1 + phi_A * alpha_A**2 * beta_A * dG2
        ds_BB = phi_C * g**3 * beta_B * dG3 + phi_B * alpha_B**2 * beta_B * dG4
        ds_AB = phi_C * f * g * (f * beta_A * dH1 * H3 + g * beta_B * H1 * dH3)
        return s_AA, s_BB, s_AB, ds_AA, ds_BB, ds_AB

    def parts(x):
        # numerator, determinant and their x-derivatives of F = N/D
        s_AA, s_BB, s_AB, d_AA, d_BB, d_AB = _corr(np.atleast_1d(np.asarray(x, dtype=float)))
        N = s_AA + s_BB + 2.0 * s_AB
        D = s_AA * s_BB - s_AB**2
        dN = d_AA + d_BB + 2.0 * d_AB
        dD = d_AA * s_BB + s_AA * d_BB - 2.0 * s_AB * d_AB
        return N, D, dN, dD

    def F(x):
        # chi-independent part of N / S(q)
        N, D, _, _ = parts(x)
        safe = D > 0
        return np.where(safe, N / np.where(safe, D, 1.0), np.inf)

    def slope_num(x):
        # sign of dF/dx (times D^2 > 0)
        N, D, dN, dD = parts(x)
        return float((dN * D - N * dD)[0])

    eps_rt = 4 * np.finfo(float).eps

    # 1) peak of S = global minimum of F on [0, 50]
    xs = np.concatenate(([0.0], np.geomspace(1e-6, 50.0, 4001)))
    Fs = F(xs)
    i = int(np.argmin(Fs))
    F0 = float(Fs[0])
    x_pk, F_min = 0.0, F0
    if i > 0:
        lo, hi = xs[i - 1], xs[min(i + 1, len(xs) - 1)]
        if not (lo == 0.0 and slope_num(0.0) >= 0.0):
            # root of dF/dx: a minimizer of F alone is limited to ~sqrt(eps) relative accuracy
            if slope_num(lo) < 0.0 < slope_num(hi):
                xm = brentq(slope_num, lo, hi, xtol=1e-300, rtol=eps_rt, maxiter=500)
            else:
                xm = float(xs[i])
            Fm = float(F(xm)[0])
            if Fm < F0:
                x_pk, F_min = float(xm), Fm
    I_pk = 1.0 / (F_min - 2.0 * chiN)

    # 2) half maximum: S = S_pk/2  <=>  F(x) = 2 F_min - 2 chiN
    target = 2.0 * F_min - 2.0 * chiN

    def h(x):
        return min(float(F(x)[0]), 1e300) - target

    # nearest crossing above the peak
    xr = np.concatenate(([x_pk], x_pk + np.geomspace(1e-6, 50.0 - x_pk, 4001)))
    hr = F(xr) - target
    j = int(np.argmax(hr >= 0.0))
    x_hi = brentq(h, xr[j - 1], xr[j], xtol=1e-300, rtol=eps_rt, maxiter=500)

    # nearest crossing below the peak
    x_lo = float("nan")
    if x_pk > 0.0:
        xl = np.concatenate(([0.0], x_pk * np.geomspace(1e-6, 1.0, 4001)))
        hl = F(xl) - target
        idx = np.where(hl >= 0.0)[0]
        if len(idx) > 0:
            k = int(idx[-1])
            x_lo = brentq(h, xl[k], xl[k + 1], xtol=1e-300, rtol=eps_rt, maxiter=500)

    return float(x_pk), float(I_pk), float(x_lo), float(x_hi)
````

## Background

**Incompressible RPA.** For a two-component incompressible melt, the RPA scattering function is
$$\frac{1}{S(q)}=\frac{S^0_{AA}+S^0_{BB}+2S^0_{AB}}{S^0_{AA}S^0_{BB}-(S^0_{AB})^2}-2\chi.$$
With $S^0_{ij}=\bar Ns_{ij}(x)$:
$$\frac{\bar N}{S(q)}=F(x)-2\chi\bar N,\qquad F=\frac{\mathcal N}{\mathcal D},\quad\mathcal N=s_{AA}+s_{BB}+2s_{AB},\quad\mathcal D=s_{AA}s_{BB}-s_{AB}^2,$$
$$(\chi\bar N)_s=\tfrac12\min_{x\ge0}F(x),\qquad x^*=\arg\min F.$$

**$q=0$ limit.** For a pure diblock, $F(0)=(Đ_Af^2+Đ_Bg^2+2fg)/[f^2g^2(Đ_AĐ_B-1)]$.
- Monodisperse ($Đ_A=Đ_B=1$): $F(0)=+\infty$, the correlation hole. Treat a zero determinant as $F=+\infty$. This recovers Leibler's $(\chi N)_s\approx10.495$, $x^*\approx3.785$.
- Polydisperse: $F(0)$ is finite. For large enough dispersity the global minimum moves to $q=0$ and a pure diblock macrophase-separates (Burger, Ruland and Semenov).
- Homopolymer blend: $\chi_s=\tfrac12[1/(\phi_AN_{w,A})+1/(\phi_BN_{w,B})]$, with weight-average lengths $N_w$.

**Competing minima.** With long homopolymers, $F(x)$ can have a local minimum at $x=0$ and another at finite $x$, and either may be global. Neither $F'(0)$ alone nor a local optimizer from a single start can decide which, so a global scan that includes $x=0$ is required.

**Precision of $x^*$.** Near a minimum, $F(x)-F_{\min}\approx\tfrac12F''(x^*)(x-x^*)^2$. Values of $F$ that agree to machine precision $\epsilon\approx2\times10^{-16}$ therefore cannot distinguish points closer than $|x-x^*|\sim\sqrt{2\epsilon F/F''}$, roughly $10^{-8}$ relative. Any method that only compares function values (golden section, Brent minimization, grid refinement) is limited to this accuracy. To reach $10^{-10}$, locate $x^*$ as a **root** of
$$\frac{dF}{dx}\propto\mathcal N'\mathcal D-\mathcal N\mathcal D',\qquad \mathcal N'=s'_{AA}+s'_{BB}+2s'_{AB},\quad\mathcal D'=s'_{AA}s_{BB}+s_{AA}s'_{BB}-2s_{AB}s'_{AB},$$
using the analytic derivatives of Subproblem 1. A sign change of this function brackets the minimum found by the global scan, and the root is simple, so Brent's method converges to near machine precision.

**Algorithm.**
1. Evaluate $F$ on a dense grid: $x=0$ plus geometrically spaced points up to $50$.
2. If the grid minimum is at $x=0$, or the slope at $0$ is non-negative when the minimum sits in the first interval, return $x^*=0$ and $\tfrac12F(0)$.
3. Otherwise, root-find $\mathcal N'\mathcal D-\mathcal N\mathcal D'$ in the bracketing interval.
4. Compare $F(x^*)$ with $F(0)$ and keep the global minimum.

## Solution

````python
import numpy as np
from scipy.optimize import brentq


def rpa_spinodal(f, phi_A, phi_B, alpha_A, alpha_B, beta_A, beta_B, D_A, D_B):
    """
    Mean-field (RPA) spinodal of a polydisperse, conformationally asymmetric A/B/A-B ternary blend.

    Inputs:
      f       : float in (0, 1), number-average A fraction of the diblock
      phi_A   : float >= 0, volume fraction of A homopolymer
      phi_B   : float >= 0, volume fraction of B homopolymer (phi_A + phi_B <= 1)
      alpha_A : float > 0, number-average N_A / N
      alpha_B : float > 0, number-average N_B / N
      beta_A  : float > 0, (b_A / b0)^2
      beta_B  : float > 0, (b_B / b0)^2
      D_A     : float in [1, 3], dispersity of every A sequence (A block and A homopolymer)
      D_B     : float in [1, 3], dispersity of every B sequence (B block and B homopolymer)

    Outputs:
      chiN_s : float, spinodal value of chi*N (N = number-average diblock length)
      x_star : float, x* = q*^2 N b0^2 / 6 of the globally most unstable mode on [0, 50];
               exactly 0.0 if the instability is at q = 0 (macrophase)
    """
    import numpy as np
    from scipy.optimize import brentq

    def _sz(y, eps):
        # Schulz-Zimm functions G, G', H, H' with eps = D - 1 >= 0:
        #   G(y) = 2[(1+eps*y)^(-1/eps) - 1 + y]/y^2,  H(y) = [1 - (1+eps*y)^(-1/eps)]/y
        y = np.asarray(y, dtype=float)
        G = np.empty_like(y); dG = np.empty_like(y)
        H = np.empty_like(y); dH = np.empty_like(y)
        sm = np.abs(y) < 0.1
        ys = y[sm]
        # series (1+eps*y)^(-1/eps) = sum_n c_n y^n, c_0 = 1, c_{n+1} = -c_n (1 + n*eps)/(n + 1)
        Gs = np.zeros_like(ys); dGs = np.zeros_like(ys)
        Hs = np.zeros_like(ys); dHs = np.zeros_like(ys)
        pw = [np.ones_like(ys)]
        for n in range(1, 82):
            pw.append(pw[-1] * ys)
        c = 1.0
        for n in range(80):
            if n >= 1:
                Hs += -c * pw[n - 1]
            if n >= 2:
                Gs += 2.0 * c * pw[n - 2]
                dHs += -(n - 1) * c * pw[n - 2]
            if n >= 3:
                dGs += 2.0 * (n - 2) * c * pw[n - 3]
            c = -c * (1.0 + n * eps) / (n + 1)
        G[sm], dG[sm], H[sm], dH[sm] = Gs, dGs, Hs, dHs
        yl = y[~sm]
        if eps == 0.0:
            T1 = np.expm1(-yl)          # (1+eps*y)^(-1/eps) - 1
            dT1 = -np.expm1(-yl)        # d/dy[(1+eps*y)^(-1/eps)] + 1
        else:
            T1 = np.expm1(-np.log1p(eps * yl) / eps)
            dT1 = -np.expm1(-(1.0 + eps) * np.log1p(eps * yl) / eps)
        Gl = 2.0 * (T1 + yl) / yl**2
        Hl = -T1 / yl
        G[~sm] = Gl
        dG[~sm] = 2.0 * dT1 / yl**2 - 2.0 * Gl / yl
        H[~sm] = Hl
        dH[~sm] = ((1.0 - dT1) - Hl) / yl
        return G, dG, H, dH

    def _corr(x_arr):
        # s_AA, s_BB, s_AB and their x-derivatives
        g = 1.0 - f
        phi_C = 1.0 - phi_A - phi_B
        eA, eB = D_A - 1.0, D_B - 1.0
        xA, xB = beta_A * x_arr, beta_B * x_arr
        G1, dG1, H1, dH1 = _sz(f * xA, eA)
        G2, dG2, _, _ = _sz(alpha_A * xA, eA)
        G3, dG3, H3, dH3 = _sz(g * xB, eB)
        G4, dG4, _, _ = _sz(alpha_B * xB, eB)
        s_AA = phi_C * f**2 * G1 + phi_A * alpha_A * G2
        s_BB = phi_C * g**2 * G3 + phi_B * alpha_B * G4
        s_AB = phi_C * f * g * H1 * H3
        ds_AA = phi_C * f**3 * beta_A * dG1 + phi_A * alpha_A**2 * beta_A * dG2
        ds_BB = phi_C * g**3 * beta_B * dG3 + phi_B * alpha_B**2 * beta_B * dG4
        ds_AB = phi_C * f * g * (f * beta_A * dH1 * H3 + g * beta_B * H1 * dH3)
        return s_AA, s_BB, s_AB, ds_AA, ds_BB, ds_AB

    def parts(x):
        # numerator, determinant and their x-derivatives of F = N/D
        s_AA, s_BB, s_AB, d_AA, d_BB, d_AB = _corr(np.atleast_1d(np.asarray(x, dtype=float)))
        N = s_AA + s_BB + 2.0 * s_AB
        D = s_AA * s_BB - s_AB**2
        dN = d_AA + d_BB + 2.0 * d_AB
        dD = d_AA * s_BB + s_AA * d_BB - 2.0 * s_AB * d_AB
        return N, D, dN, dD

    def F(x):
        N, D, _, _ = parts(x)
        safe = D > 0
        return np.where(safe, N / np.where(safe, D, 1.0), np.inf)

    def slope_num(x):
        # sign of dF/dx (times D^2 > 0)
        N, D, dN, dD = parts(x)
        return float((dN * D - N * dD)[0])

    # 1) global scan including the q = 0 endpoint (F can have two competing minima)
    xs = np.concatenate(([0.0], np.geomspace(1e-6, 50.0, 4001)))
    Fs = F(xs)
    i = int(np.argmin(Fs))
    F0 = float(Fs[0])
    if i == 0:
        return float(0.5 * F0), 0.0
    lo, hi = xs[i - 1], xs[min(i + 1, len(xs) - 1)]
    if lo == 0.0 and slope_num(0.0) >= 0.0:
        return float(0.5 * F0), 0.0

    # 2) locate the minimum as the root of dF/dx (a minimizer of F alone is limited
    #    to ~sqrt(machine eps) relative accuracy because F is quadratic at its minimum)
    if slope_num(lo) < 0.0 < slope_num(hi):
        x_star = brentq(slope_num, lo, hi, xtol=1e-300, rtol=4 * np.finfo(float).eps, maxiter=500)
    else:
        x_star = float(xs[i])
    F_min = float(F(x_star)[0])
    if F0 <= F_min:
        return float(0.5 * F0), 0.0
    return float(0.5 * F_min), float(x_star)
````

## Background

**Lifshitz line.** In the RPA, $\bar N/S(q)=F(x)-2\chi\bar N$, so $\chi$ enters only as a constant shift. Expand $F=F_0+F_1x+O(x^2)$. The Lifshitz line is $F_1=0$: the $q^2$ coefficient of the Landau–Ginzburg free energy vanishes, and the peak of the disordered-state $S(q)$ leaves $q=0$ (Holyst and Schick; Broseta and Fredrickson). Because $F_1$ does not depend on $\chi$, $\phi_L$ depends only on molecular parameters. When the $q=0$ instability is reached continuously, the Lifshitz line meets the spinodal at the isotropic Lifshitz point (Bates et al.).

**Small-$x$ expansion with polydispersity and conformational asymmetry.** From the series in Subproblem 1:
$$G_\varepsilon(y)=Đ-\tfrac13Đ(2Đ-1)\,y+O(y^2),\qquad H_\varepsilon(y)=1-\tfrac12Đ\,y+O(y^2).$$
With $g=1-f$, $a=\phi_A\alpha_A$, $b=\phi_B\alpha_B$ and $c=\phi_C$, the correlation functions are $s_{ij}=s^{(0)}_{ij}+s^{(1)}_{ij}x+O(x^2)$, where
$$s^{(0)}_{AA}=Đ_A(a+cf^2),\quad s^{(0)}_{BB}=Đ_B(b+cg^2),\quad s^{(0)}_{AB}=cfg,$$
$$s^{(1)}_{AA}=-\tfrac13\beta_AĐ_A(2Đ_A-1)(a\alpha_A+cf^3),\quad s^{(1)}_{BB}=-\tfrac13\beta_BĐ_B(2Đ_B-1)(b\alpha_B+cg^3),$$
$$s^{(1)}_{AB}=-\tfrac12cfg\,(Đ_Af\beta_A+Đ_Bg\beta_B).$$
With $\mathcal N=s_{AA}+s_{BB}+2s_{AB}$ and $\mathcal D=s_{AA}s_{BB}-s_{AB}^2$, we have $F_1=(\mathcal N_1\mathcal D_0-\mathcal N_0\mathcal D_1)/\mathcal D_0^2$. The Lifshitz condition is the polynomial root
$$\Lambda(\phi_H)\equiv\mathcal N_1\mathcal D_0-\mathcal N_0\mathcal D_1=0,\qquad(\chi\bar N)_L=\frac{\mathcal N_0}{2\mathcal D_0}\Big|_{\phi_L}.$$
Bracket the root on a grid and refine it with Brent's method.

Two shortcuts fail the accuracy requirement:
- Finite differences of $F$ at $x=0$ carry an $O(h)$ bias.
- Tracking where $x^*\to0$ is limited by the quartic flatness of $F$ and fails when minima compete.

For strongly polydisperse diblocks, $\Lambda(0)$ itself can become positive, because a pure diblock can then macrophase-separate. The task restricts parameters to cases with a single sign change.

**Symmetric monodisperse check.** For $f=1/2$, $\alpha_A=\alpha_B=\alpha$, $r=1/2$, $\beta_A=\beta_B$ and $Đ_A=Đ_B=1$:
$$\phi_L=\frac{1}{1+2\alpha^2},\qquad(\chi N)_L=\frac{2(1+2\alpha^2)}{\alpha}.$$

**Lifshitz line vs. spinodal character.** The sign of $F_1$ is *local* information at $q=0$. With long homopolymers, $F$ can have a local minimum at $x=0$ and a deeper one at finite $x$, so the melt still microphase-separates although $\phi_H>\phi_L$. The reverse can also occur. The spinodal and the micro/macro label must therefore come from the **global** minimization of Subproblem 2 at each $\phi_H$, not from comparing $\phi_H$ with $\phi_L$.

## Solution

````python
import numpy as np
from scipy.optimize import brentq


def lifshitz_isopleth(f, alpha_A, alpha_B, beta_A, beta_B, D_A, D_B, r, phi_H_grid):
    """
    Lifshitz-line composition and RPA spinodal along an isopleth of a
    polydisperse, conformationally asymmetric A/B/A-B ternary blend.

    Inputs:
      f          : float in (0, 1), number-average A fraction of the diblock
      alpha_A    : float > 0, number-average N_A / N
      alpha_B    : float > 0, number-average N_B / N
      beta_A     : float > 0, (b_A / b0)^2
      beta_B     : float > 0, (b_B / b0)^2
      D_A        : float >= 1, dispersity of every A sequence (A block and A homopolymer)
      D_B        : float >= 1, dispersity of every B sequence (B block and B homopolymer)
      r          : float in (0, 1), phi_A / (phi_A + phi_B)
      phi_H_grid : 1D array-like of total homopolymer fractions in (0, 1)

    Outputs:
      dict with keys
        'phi_L'  : float, homopolymer fraction where dF/dx at x = 0 changes sign
        'chiN_L' : float, F(0)/2 evaluated at phi_L
        'chiN_s' : np.ndarray, spinodal chi*N at each phi_H in phi_H_grid
        'x_star' : np.ndarray, x* = q*^2 N b0^2 / 6 of the globally unstable mode (0.0 for macrophase)
        'phase'  : list of str, 'micro' (x_star > 0) or 'macro' (x_star == 0) for each phi_H
    """
    import numpy as np
    from scipy.optimize import brentq

    def rpa_spinodal(f, phi_A, phi_B, alpha_A, alpha_B, beta_A, beta_B, D_A, D_B):
        """Mean-field RPA spinodal (chiN_s, x_star); x_star = 0.0 for macrophase."""
        import numpy as np
        from scipy.optimize import minimize_scalar

        def _sz_pow(y, eps):
            # T(y) = (1 + eps*y)^(-1/eps) - 1, computed without cancellation; eps = D - 1 >= 0
            if eps == 0.0:
                return np.expm1(-y)
            return np.expm1(-np.log1p(eps * y) / eps)

        def _G(y, eps):
            # Schulz-Zimm-averaged Debye function 2[(1+eps*y)^(-1/eps) - 1 + y]/y^2, G(0) = 1 + eps
            y = np.asarray(y, dtype=float)
            out = np.empty_like(y)
            small = np.abs(y) < 1e-2
            ys = y[small]
            # series: (1+eps*y)^(-1/eps) = sum_n c_n y^n, c_{n+1} = -c_n (1 + n*eps)/(n + 1)
            c = np.full_like(ys, 0.5 * (1.0 + eps))
            s = np.zeros_like(ys)
            p = np.ones_like(ys)
            for n in range(2, 30):
                s += c * p
                c = -c * (1.0 + n * eps) / (n + 1)
                p = p * ys
            out[small] = 2.0 * s
            yl = y[~small]
            out[~small] = 2.0 * (_sz_pow(yl, eps) + yl) / yl**2
            return out

        def _H(y, eps):
            # Schulz-Zimm-averaged block function [1 - (1+eps*y)^(-1/eps)]/y, H(0) = 1
            y = np.asarray(y, dtype=float)
            out = np.ones_like(y)
            nz = y != 0
            out[nz] = -_sz_pow(y[nz], eps) / y[nz]
            return out

        def F(x):
            # chi-independent part of N / S(q): F = (sAA + sBB + 2 sAB) / (sAA sBB - sAB^2)
            x_arr = np.atleast_1d(np.asarray(x, dtype=float))
            g = 1.0 - f
            phi_C = 1.0 - phi_A - phi_B
            eA, eB = D_A - 1.0, D_B - 1.0
            xA, xB = beta_A * x_arr, beta_B * x_arr
            s_AA = phi_C * f**2 * _G(f * xA, eA) + phi_A * alpha_A * _G(alpha_A * xA, eA)
            s_BB = phi_C * g**2 * _G(g * xB, eB) + phi_B * alpha_B * _G(alpha_B * xB, eB)
            s_AB = phi_C * f * g * _H(f * xA, eA) * _H(g * xB, eB)
            num = s_AA + s_BB + 2.0 * s_AB
            den = s_AA * s_BB - s_AB**2
            safe = den > 0
            return np.where(safe, num / np.where(safe, den, 1.0), np.inf)

        x_max = 50.0
        # global scan including the q = 0 endpoint (F can have two competing minima)
        xs = np.concatenate(([0.0], np.geomspace(1e-6, x_max, 4001)))
        Fs = F(xs)
        i = int(np.argmin(Fs))
        lo, hi = xs[max(i - 1, 0)], xs[min(i + 1, len(xs) - 1)]

        # local refinement around the global grid minimum
        res = minimize_scalar(lambda t: float(F(t)[0]), bounds=(lo, hi), method="bounded",
                              options={"xatol": 1e-12, "maxiter": 1000})

        F0 = float(F(0.0)[0])
        cands = [(float(res.x), float(res.fun)), (0.0, F0), (float(xs[i]), float(Fs[i]))]
        x_star, F_min = min(cands, key=lambda c: c[1])
        if x_star < 1e-8:
            x_star, F_min = 0.0, F0
        return float(0.5 * F_min), float(x_star)

    g = 1.0 - f

    def coeffs(phi_H):
        # O(x^0) and O(x^1) coefficients of the numerator and determinant of F(x)
        # (Schulz-Zimm: G(y) = D - D(2D - 1) y/3 + ..., H(y) = 1 - D y/2 + ...)
        a = r * phi_H * alpha_A
        b = (1.0 - r) * phi_H * alpha_B
        c = 1.0 - phi_H
        sAA0 = D_A * (a + c * f**2)
        sBB0 = D_B * (b + c * g**2)
        sAB0 = c * f * g
        sAA1 = -beta_A * D_A * (2.0 * D_A - 1.0) * (a * alpha_A + c * f**3) / 3.0
        sBB1 = -beta_B * D_B * (2.0 * D_B - 1.0) * (b * alpha_B + c * g**3) / 3.0
        sAB1 = -0.5 * c * f * g * (D_A * f * beta_A + D_B * g * beta_B)
        N0, N1 = sAA0 + sBB0 + 2.0 * sAB0, sAA1 + sBB1 + 2.0 * sAB1
        D0 = sAA0 * sBB0 - sAB0**2
        D1 = sAA1 * sBB0 + sAA0 * sBB1 - 2.0 * sAB0 * sAB1
        return N0, N1, D0, D1

    def lifshitz_fn(phi_H):
        # proportional to dF/dx at x = 0 (positive factor 1/D0^2 removed)
        N0, N1, D0, D1 = coeffs(phi_H)
        return N1 * D0 - N0 * D1

    # bracket the single sign change, then refine with Brent's method
    pg = np.linspace(1e-9, 1.0 - 1e-9, 20001)
    Lg = np.array([lifshitz_fn(p) for p in pg])
    k = np.where(np.sign(Lg[:-1]) != np.sign(Lg[1:]))[0][0]
    phi_L = brentq(lifshitz_fn, pg[k], pg[k + 1], xtol=1e-15, rtol=4 * np.finfo(float).eps)
    N0, _, D0, _ = coeffs(phi_L)
    chiN_L = 0.5 * N0 / D0

    # spinodal along the isopleth; phase label from the GLOBAL minimizer
    chiN_s, x_star, phase = [], [], []
    for p in np.atleast_1d(np.asarray(phi_H_grid, dtype=float)):
        c_, x_ = rpa_spinodal(f, r * p, (1.0 - r) * p, alpha_A, alpha_B, beta_A, beta_B, D_A, D_B)
        chiN_s.append(c_)
        x_star.append(x_)
        phase.append("micro" if x_ > 0.0 else "macro")

    return {"phi_L": float(phi_L), "chiN_L": float(chiN_L),
            "chiN_s": np.array(chiN_s), "x_star": np.array(x_star), "phase": phase}
````
